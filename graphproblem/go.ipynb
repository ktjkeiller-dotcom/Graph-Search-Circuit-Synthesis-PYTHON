```python
from graphproblem.ringenum import Ring
from graphproblem.heuristics.vheuristic import VHeuristic
from graphproblem.searchmethods import astar
```

```python
RING = Ring.CYCLOTOMIC2
heuristic = VHeuristic()
search_method = astar.AStar(heuristic, RING)
```

```python
gates = RING.ring_gates

CS = gates["CH-"]
EYE = gates["I-"]
HI = gates["H-"] @ EYE
IH = EYE @ gates["H-"]
SI = gates["Z-"] @ EYE
CX = gates["CX-"]

TEST = CS*CX*HI*SI*CS*IH*CX*CS*IH*CS*CX
```

```python
seq, V  = search_method.search(TEST)
```
